```python
import pandas as pd
import numpy as np

# ==========================================
# PHASE 11 — FORECASTING
# ==========================================

# Load processed dataset
df = pd.read_csv(
    "../data/processed/uac_features.csv",
    parse_dates=["Date"]
)

# Sort chronologically
df = df.sort_values("Date").reset_index(drop=True)

# Make sure total_system_load exists
if "total_system_load" not in df.columns:
    df["total_system_load"] = (
        df["Children in CBP custody"] +
        df["Children in HHS Care"]
    )

# Create forecasting dataset
forecast_df = (
    df[["Date", "total_system_load"]]
    .dropna()
    .copy()
)

# Sort by date again
forecast_df = forecast_df.sort_values("Date").reset_index(drop=True)

# Display the data
print("Forecasting dataset created successfully!")
print("Number of observations:", len(forecast_df))
print("Start date:", forecast_df["Date"].min())
print("End date:", forecast_df["Date"].max())

forecast_df.head()
```

```python
# Chronological train-test split
split_index = int(len(forecast_df) * 0.80)

train = forecast_df.iloc[:split_index].copy()
test = forecast_df.iloc[split_index:].copy()

print("Training observations:", len(train))
print("Testing observations:", len(test))
```

```python
print("Training period:")
print(train["Date"].min(), "to", train["Date"].max())

print("\nTesting period:")
print(test["Date"].min(), "to", test["Date"].max())
```

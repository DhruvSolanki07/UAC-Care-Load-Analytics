```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv(
    "../data/processed/uac_features.csv",
    parse_dates=["Date"]
)
```

```python
fig = px.line(
    df,
    x="Date",
    y="total_system_load",
    title="Total System Load Over Time"
)

fig.show()
```

```python
fig = px.line(
    df,
    x="Date",
    y=[
        "Children in CBP custody",
        "Children in HHS Care"
    ],
    title="CBP Custody vs HHS Care"
)

fig.show()
```

```python
fig = px.line(
    df,
    x="Date",
    y=[
        "Children transferred out of CBP custody",
        "Children discharged from HHS Care"
    ],
    title="Transfers vs Discharges"
)

fig.show()
```

```python
fig = px.bar(
    df,
    x="Date",
    y="net_intake",
    title="Net Intake Pressure"
)

fig.show()
```

```python
fig = px.line(
    df,
    x="Date",
    y="daily_load_change",
    title="Daily Change in Total System Load"
)

fig.show()
```

```python
print(df.columns.tolist())
```

```python
df["total_system_load"] = (
    df["Children in CBP custody"] +
    df["Children in HHS Care"]
)

df["load_7d_avg"] = (
    df["total_system_load"]
    .rolling(window=7, min_periods=1)
    .mean()
)

df["load_14d_avg"] = (
    df["total_system_load"]
    .rolling(window=14, min_periods=1)
    .mean()
)
```

```python
import plotly.express as px

fig = px.line(
    df,
    x="Date",
    y=[
        "total_system_load",
        "load_7d_avg",
        "load_14d_avg"
    ],
    title="System Load and Rolling Trends"
)

fig.show()
```

```python
monthly = (
    df.set_index("Date")
      .resample("ME")
      .agg({
          "total_system_load": "mean",
          "net_intake": "sum",
          "Children in CBP custody": "mean",
          "Children in HHS Care": "mean"
      })
      .reset_index()
)
```

```python
fig = px.line(
    monthly,
    x="Date",
    y="total_system_load",
    title="Monthly Average Observed System Load"
)

fig.show()
```

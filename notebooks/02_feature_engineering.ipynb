```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/processed/uac_cleaned.csv",
    parse_dates=["Date"]
)
```

```python
df["total_system_load"] = (
    df["Children in CBP custody"] +
    df["Children in HHS Care"]
)
```

```python
df["net_intake"] = (
    df["Children transferred out of CBP custody"]
    -
    df["Children discharged from HHS Care"]
)
```

```python
df["daily_load_change"] = (
    df["total_system_load"].diff()
)
```

```python
df["care_load_growth_rate"] = (
    df["total_system_load"].pct_change() * 100
)
```

```python
df["care_load_growth_rate"] = (
    df["total_system_load"].pct_change() * 100
)
```

```python
df["load_7d_avg"] = (
    df["total_system_load"]
    .rolling(7, min_periods=3)
    .mean()
)
```

```python
df["net_intake_7obs"] = (
    df["net_intake"]
    .rolling(7, min_periods=3)
    .sum()
)
```

```python
df["load_volatility_7obs"] = (
    df["daily_load_change"]
    .rolling(7, min_periods=3)
    .std()
)
```

```python
df["discharge_offset_ratio"] = np.where(
    df["Children transferred out of CBP custody"] > 0,
    df["Children discharged from HHS Care"] /
    df["Children transferred out of CBP custody"],
    np.nan
)
```

```python
df.to_csv(
    "../data/processed/uac_features.csv",
    index=False
)
```
